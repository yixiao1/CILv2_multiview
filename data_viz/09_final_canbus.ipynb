```python
def decode_float_directions_to_str(float_direction):
    if float_direction == 1.0:
        return 'Turn Left'
    elif float_direction == 2.0:
        return 'Turn Right'
    elif float_direction == 3.0:
        return 'Go Straight'
    elif float_direction == 4.0 or float_direction == 5.0 or float_direction == 6.0 or float_direction == 0.0:
        return 'Follow Lane'
    # elif float_direction == 5.0:
    #     return 'CHANGE LANE LEFT'
    # elif float_direction == 6.0:
    #     return 'CHANGE LANE RIGHT'
    else:
        return f'Unknown ({float_direction})'
```

```python
# Add a column with the decoded directions
df['Command'] = df['direction'].apply(decode_float_directions_to_str)
```

```python
# Add a column on the magnitude of acceleration and G-forces
df['acceleration_magnitude'] = (df['accel_x']**2 + df['accel_y']**2 + df['accel_z']**2)**0.5
df['GForce'] = df['acceleration_magnitude'] / 9.81
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from typing import Union
import numpy as np

def plot_route_2d_multi_user(df, town: str, route: str, tag: str = None, size: Union[float, str] = 10, cmap: str = 'viridis'):
    # Select data for the specified route
    df_town = df[df.town == town]
    df_route = df_town[df_town.route == route]
    
    if df_route.empty:
        print(f"No data found for route {route}")
        return

    if tag is not None:
        df_route = df_route[df_route.tag == tag]
    
    # Set up the plot style
    plt.figure(figsize=(15, 10))
    sns.set_style("darkgrid")

    # Create a color map for users
    users = df_route['user'].unique()
    n_users = len(users)
    color_map = plt.cm.get_cmap(cmap)
    color_norm = plt.Normalize(vmin=0, vmax=n_users - 1)

    # Plot each user's route
    for i, user in enumerate(users):
        df_user = df_route[df_route.user == user]
        
        if isinstance(size, str):
            size_values = df_user.get(size)
        else:
            size_values = size

        color = color_map(color_norm(i))
        
        plt.scatter(df_user['pos_x'], -df_user['pos_y'], 
                    c=[color], 
                    s=size_values, alpha=0.05, label=user)

    # Set labels and title
    plt.xlabel('X Position')
    plt.ylabel('Y Position')
    title = f"2D Route {route} for All Users"
    title = title + f" ({tag})" if tag is not None else title
    plt.title(title)

    # Add colorbar
    sm = plt.cm.ScalarMappable(cmap=color_map, norm=color_norm)
    sm.set_array([])
    cbar = plt.colorbar(sm)
    cbar.set_label('Users', rotation=270, labelpad=15)
    
    # Set tick labels on colorbar to user names
    cbar.set_ticks(np.linspace(0, n_users - 1, n_users))
    cbar.set_ticklabels(users)

    plt.tight_layout()
    plt.show()

# Usage example
# plot_route_2d_multi_user(df, route='R01', size='speed', cmap='viridis')
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_high_acceleration_events(df, user, route, town, threshold=2.0):
    route_data = df[(df['user'] == user) & (df['route'] == route) & (df['town'] == town)].copy()

    # Get the total acceleration from the accelerometer data
    high_accel = route_data[route_data['gforce_magnitude'] > threshold]
    
    plt.figure(figsize=(12, 8))
    plt.scatter(route_data['pos_x'], route_data['pos_y'], 
                c='lightgray', s=5, label='Route')
    plt.scatter(high_accel['pos_x'], high_accel['pos_y'], 
                c='red', s=20, label=f'Acceleration > {threshold}')
    plt.title(f"High Acceleration Events for User {user} on Route {route}")
    plt.xlabel('X Position')
    plt.ylabel('Y Position')
    plt.legend()
    plt.axis('equal')
    plt.show()

# Usage
plot_high_acceleration_events(df, 30, 'R01', 'Town01',threshold=10.0)
```

```python
def plot_high_acceleration_events(df, user, town, route, threshold_g=2.0):
    route_data = df[(df['user'] == user) & (df['route'] == route) & (df['town'] == town)].copy()
    # Get the total acceleration from the accelerometer data
    route_data['total_acceleration_g'] = np.sqrt(route_data['accel_x'] ** 2 + 
                                               route_data['accel_y'] ** 2 + 
                                               route_data['accel_z'] ** 2) / 9.81
    
    high_accel = route_data[route_data['total_acceleration_g'] > threshold_g]
    
    plt.figure(figsize=(12, 8))
    plt.scatter(route_data['pos_x'], route_data['pos_y'], 
                c='lightgray', s=5, label='Route')
    scatter = plt.scatter(high_accel['pos_x'], high_accel['pos_y'], 
                c=high_accel['total_acceleration_g'], cmap='magma', s=20, 
                label=f'Acceleration > {threshold_g} G')
    plt.colorbar(scatter, label='G-force')
    plt.title(f"High Acceleration Events for User {user} on Route {route}")
    plt.xlabel('X Position')
    plt.ylabel('Y Position')
    plt.legend()
    plt.axis('equal')
    plt.show()


plot_high_acceleration_events(df, 19, 'Town01', 'R01', threshold_g=2.0)
```
